```python
# credit card transactor validator 15
def validate_transaction(balance,amount,daily_limit,used_today):
  if amount>0 and amount<=balance and amount<=daily_limit:
    rem_balance = balance - amount
    rem_limit = daily_limit - amount - used_today
    if rem_balance>=1000:
      print("Transaction Approve-->=")
      print("Remaining Balance-->=",rem_balance)
      print("Remainig Limit-->=",rem_limit)
    else:
      print("Transaction Failed")
      print("Reason---->  Balance less than 1000")
validate_transaction(50000,10000,20000,5000)


```
